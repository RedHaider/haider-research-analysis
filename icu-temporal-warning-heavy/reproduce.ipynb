# Early warning from simulated vital signs

Compare gradient boosting with eight last-measured vitals against the same model with per-vital averages, variability, changes between time windows and observation counts. Report AUROC, average precision and Brier score on stratified 25% patient holdouts. Generate 3,200 independent simulated patient trajectories per seed. Event probability depends on a latent level and temporal drift.

```python
from pathlib import Path
import subprocess, sys
root = Path.cwd() / 'icu-temporal-warning-heavy-run'
root.mkdir(exist_ok=True)
subprocess.run([sys.executable,'-m','pip','install',*['numpy==2.2.6', 'scikit-learn==1.7.2', 'scipy==1.15.3', 'joblib==1.6.0']],check=True)
```

```python
(root / 'train.py').write_text('import json\nimport os\nfrom pathlib import Path\n\nimport joblib\nimport numpy as np\nfrom sklearn.metrics import (\n    average_precision_score,\n    brier_score_loss,\n    roc_auc_score,\n)\n\nROOT = Path(__file__).resolve().parent\nDATA = Path(os.environ.get("PORTFOLIO_DATA", ROOT / "data"))\nSEEDS = [13, 37, 61]\n\n\ndef save(obj, name):\n    folder = ROOT / "checkpoints"\n    folder.mkdir(exist_ok=True)\n    joblib.dump(obj, folder / (name + ".joblib"), compress=3)\n\n\ndef scores(y, p):\n    return {\n        "auroc": float(roc_auc_score(y, p)),\n        "average_precision": float(average_precision_score(y, p)),\n    }\n\n\nfrom sklearn.ensemble import HistGradientBoostingClassifier\nfrom sklearn.metrics import brier_score_loss\nfrom sklearn.model_selection import train_test_split\n\nVITALS = ["HR", "RespRate", "Temp", "SysABP", "SpO2", "MAP", "NISysABP", "SaO2"]\n\n\ndef summarize(series):\n    static = []\n    dynamic = []\n    for name in VITALS:\n        values = sorted(series.get(name, []))\n        if values:\n            times, y = np.array(values).T\n            static.append(y[-1])\n            early = y[times < 12]\n            late = y[times >= 12]\n            change = late.mean() - early.mean() if len(early) and len(late) else np.nan\n            dynamic.extend([y.mean(), y.std(), change, np.log1p(len(y))])\n        else:\n            static.append(np.nan)\n            dynamic.extend([np.nan, np.nan, np.nan, 0])\n    return static + dynamic\n\n\ndef simulated(seed):\n    from scipy.special import expit\n\n    rng = np.random.default_rng(seed)\n    x = []\n    y = []\n    for _ in range(3200):\n        base, slope = rng.normal(size=2)\n        y.append(rng.random() < expit(-1.6 + 0.45 * base + 1.2 * slope))\n        series = {}\n        for j, k in enumerate(VITALS):\n            t = np.sort(rng.uniform(0, 24, rng.integers(2, 19)))\n            v = (\n                60\n                + 4 * j\n                + 3 * base\n                + (t / 24 - 0.5) * 5 * slope * (-1) ** j\n                + rng.normal(0, 3, len(t))\n            )\n            series[k] = list(zip(t, v))\n        x.append(summarize(series))\n    return np.array(x), np.array(y)\n\n\ndef run(seed):\n    x, y = simulated(seed)\n    train, test = train_test_split(\n        np.arange(len(y)), test_size=0.25, stratify=y, random_state=seed\n    )\n    out = {}\n    for method, k in [("last_values", 8), ("temporal_summaries", 40)]:\n        model = HistGradientBoostingClassifier(\n            max_iter=125,\n            min_samples_leaf=24,\n            max_leaf_nodes=15,\n            l2_regularization=2,\n            random_state=seed,\n        )\n        model.fit(x[train, :k], y[train])\n        p = model.predict_proba(x[test, :k])[:, 1]\n        out[method] = {\n            **scores(y[test], p),\n            "brier": float(brier_score_loss(y[test], p)),\n        }\n        save((model, test), f"{method}-{seed}")\n    return out\n\n\nif __name__ == "__main__":\n    runs = {s: run(s) for s in SEEDS}\n    first = next(iter(runs.values()))\n    means = {\n        k: {m: float(np.mean([r[k][m] for r in runs.values()])) for m in v}\n        for k, v in first.items()\n    }\n    (ROOT / "results.json").write_text(\n        json.dumps({"seeds": SEEDS, "runs": runs, "means": means}, indent=2)\n    )\n    print(means)\n',encoding='utf8')
subprocess.run([sys.executable,'train.py'],cwd=root,check=True)
```
