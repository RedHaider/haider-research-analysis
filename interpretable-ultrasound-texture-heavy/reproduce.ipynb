# Interpretable ultrasound texture classification (simulation)

Compare raw pixels, 54 interpretable texture summaries, and their combination using the same regularized classifier. The summaries measure intensities, residuals and gradient magnitudes at three spatial scales across the full image and central crop. Repeat with three stratified 85% training subsamples. Each seed generates 1,800 training and 600 held-out phantoms. The generator varies texture and elongation by class. It supplies a controlled test of feature sensitivity.

```python
from pathlib import Path
import subprocess, sys
root = Path.cwd() / 'interpretable-ultrasound-texture-heavy-run'
root.mkdir(exist_ok=True)
subprocess.run([sys.executable,'-m','pip','install',*['numpy==2.2.6', 'scipy==1.15.3', 'scikit-learn==1.7.2', 'joblib==1.6.0']],check=True)
```

```python
(root / 'train.py').write_text('import json\nimport os\nfrom pathlib import Path\n\nimport joblib\nimport numpy as np\nfrom scipy import ndimage as ndi\nfrom sklearn.linear_model import LogisticRegression\nfrom sklearn.metrics import balanced_accuracy_score, roc_auc_score\nfrom sklearn.model_selection import train_test_split\nfrom sklearn.pipeline import make_pipeline\nfrom sklearn.preprocessing import StandardScaler\n\nROOT = Path(__file__).resolve().parent\nDATA = Path(os.environ.get("PORTFOLIO_DATA", str(ROOT / "data")))\nSEEDS = [13, 37, 61]\n\n\ndef classifier():\n    return make_pipeline(StandardScaler(), LogisticRegression(C=0.05, max_iter=700))\n\n\ndef save(model, name):\n    path = ROOT / "checkpoints"\n    path.mkdir(exist_ok=True)\n    joblib.dump(model, path / (name + ".joblib"), compress=3)\n\n\ndef phantom(n, seed):\n    rng = np.random.default_rng(seed)\n    y = rng.integers(2, size=n)\n    u, v = np.mgrid[-1:1:28j, -1:1:28j]\n    x = []\n    for label in y:\n        a, b = rng.normal(0, 0.22, 2)\n        scale = rng.uniform(0.14, 0.38)\n        field = np.exp(\n            -(((u - a) / (scale * (1 + 0.55 * label))) ** 2 + ((v - b) / scale) ** 2)\n            / 2\n        )\n        grain = ndi.gaussian_filter(rng.normal(size=u.shape), 0.5 + 0.4 * label)\n        x.append(np.clip(0.30 + field * rng.uniform(0.20, 0.5) + grain * 0.16, 0, 1))\n    return np.asarray(x), y\n\n\ndef image_data(seed, filename):\n    return [phantom(n, seed + offset) for n, offset in [(1800, 0), (600, 100)]]\n\n\ndef evaluate(y, p):\n    return {\n        "auroc": float(roc_auc_score(y, p)),\n        "balanced_accuracy": float(balanced_accuracy_score(y, p >= 0.5)),\n    }\n\n\ndef descriptors(x):\n    parts = []\n    for region in [x, x[:, 7:21, 7:21]]:\n        for sigma in [0.7, 1.5, 3]:\n            smooth = ndi.gaussian_filter(region, (0, sigma, sigma))\n            residual = np.abs(region - smooth)\n            edge = np.hypot(np.gradient(smooth, axis=1), np.gradient(smooth, axis=2))\n            for field in [smooth, residual, edge]:\n                parts.append(np.quantile(field, [0.1, 0.5, 0.9], axis=(1, 2)).T)\n    return np.hstack(parts)\n\n\ndef run(seed):\n    (x, y), (z, t) = image_data(seed, "breastmnist.npz")\n    tr, _ = train_test_split(\n        np.arange(len(x)), train_size=0.85, stratify=y, random_state=seed\n    )\n    a, b = descriptors(x[tr]), descriptors(z)\n    out = {}\n    for name, f, g in [\n        ("pixels", x[tr].reshape(len(tr), -1), z.reshape(len(z), -1)),\n        ("texture", a, b),\n        (\n            "combined",\n            np.c_[a, x[tr].reshape(len(tr), -1)],\n            np.c_[b, z.reshape(len(z), -1)],\n        ),\n    ]:\n        model = classifier().fit(f, y[tr])\n        save(model, f"{name}-{seed}")\n        out[name] = evaluate(t, model.predict_proba(g)[:, 1])\n    return out\n\n\nif __name__ == "__main__":\n    runs = {s: run(s) for s in SEEDS}\n    first = next(iter(runs.values()))\n    means = {\n        k: {m: float(np.mean([r[k][m] for r in runs.values()])) for m in v}\n        for k, v in first.items()\n    }\n    (ROOT / "results.json").write_text(\n        json.dumps({"seeds": SEEDS, "runs": runs, "means": means}, indent=2)\n    )\n    print(means)\n',encoding='utf8')
subprocess.run([sys.executable,'train.py'],cwd=root,check=True)
```
