# Sleepâ€“wake classification with sparse labels (public data)

Hold out one quarter of participants. Train a 96-tree forest with 10% of the training epochs labeled. Retrain after adding training-only pseudo-labels with probability below 0.08 or above 0.92. Metrics pool held-out epochs within each split. PhysioNet provides Apple Watch heart-rate samples and polysomnography stage labels. Aggregate heart-rate level and variability over two windows ending at the end of each 30-second epoch. Drop epochs without enough aligned readings. The target is sleep versus wake. The experiment uses heart rate alone.

```python
from pathlib import Path
import subprocess, sys
root = Path.cwd() / 'label-efficient-sleep-real-data-run'
root.mkdir(exist_ok=True)
subprocess.run([sys.executable,'-m','pip','install',*['numpy==2.2.6', 'scipy==1.15.3', 'scikit-learn==1.7.2', 'joblib==1.6.0']],check=True)
```

```python
(root / 'prepare.py').write_text('import hashlib\nimport os\nfrom pathlib import Path\n\nDATA = Path(os.environ.get("PORTFOLIO_DATA", Path(__file__).resolve().parent / "data"))\nfrom concurrent.futures import ThreadPoolExecutor\nfrom urllib.request import urlopen\n\nbase = "https://physionet.org/files/sleep-accel/1.0.0/"\nmanifest = urlopen(base + "SHA256SUMS.txt").read().decode().splitlines()\n\n\ndef fetch(line):\n    checksum, name = line.split()\n    name = name.lstrip("*").removeprefix("./")\n    if not name.startswith(("heart_rate/", "labels/")):\n        return\n    p = DATA / "sleep-accel" / name\n    p.parent.mkdir(parents=True, exist_ok=True)\n    if not p.exists():\n        p.write_bytes(urlopen(base + name, timeout=90).read())\n    assert hashlib.sha256(p.read_bytes()).hexdigest() == checksum\n\n\nwith ThreadPoolExecutor(4) as pool:\n    list(pool.map(fetch, manifest))\n',encoding='utf8')
subprocess.run([sys.executable,'prepare.py'],cwd=root,check=True)
```

```python
(root / 'train.py').write_text('import json\nimport os\nfrom pathlib import Path\n\nimport joblib\nimport numpy as np\nfrom sklearn.metrics import (\n    average_precision_score,\n    balanced_accuracy_score,\n    roc_auc_score,\n)\n\nROOT = Path(__file__).resolve().parent\nDATA = Path(os.environ.get("PORTFOLIO_DATA", ROOT / "data"))\nSEEDS = [13, 37, 61]\n\n\ndef save(obj, name):\n    folder = ROOT / "checkpoints"\n    folder.mkdir(exist_ok=True)\n    joblib.dump(obj, folder / (name + ".joblib"), compress=3)\n\n\ndef scores(y, p):\n    return {\n        "auroc": float(roc_auc_score(y, p)),\n        "average_precision": float(average_precision_score(y, p)),\n    }\n\n\ndef dataset(seed):\n    x = []\n    y = []\n    g = []\n    for file in sorted((DATA / "sleep-accel/labels").glob("*sleep.txt")):\n        pid = file.name.split("_")[0]\n        hrfile = DATA / "sleep-accel/heart_rate" / (pid + "_heartrate.txt")\n        if not hrfile.exists():\n            continue\n        hr = np.loadtxt(hrfile, delimiter=",")\n        hr = hr[np.argsort(hr[:, 0])]\n        for stamp, stage in np.loadtxt(file):\n            if stage < 0:\n                continue\n            idx = np.searchsorted(hr[:, 0], [stamp - 300, stamp - 60, stamp + 30])\n            long = hr[idx[0] : idx[2], 1]\n            short = hr[idx[1] : idx[2], 1]\n            if len(short) < 2 or len(long) < 4:\n                continue\n            x.append(\n                [\n                    short.mean(),\n                    short.std(),\n                    long.mean(),\n                    long.std(),\n                    short.mean() - long.mean(),\n                    np.ptp(long),\n                    np.median(long),\n                    short[-1] - short[0],\n                    len(short),\n                    len(long),\n                ]\n            )\n            y.append(int(stage != 0))\n            g.append(pid)\n    assert len(set(g)) >= 15\n    return np.array(x), np.array(y), np.array(g)\n\n\nfrom sklearn.ensemble import ExtraTreesClassifier\nfrom sklearn.model_selection import GroupShuffleSplit\n\n\ndef run(seed):\n    x, y, g = dataset(seed)\n    tr, te = next(\n        GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=seed).split(x, y, g)\n    )\n    rng = np.random.default_rng(seed)\n    labeled = rng.choice(tr, int(0.10 * len(tr)), replace=False)\n    unlabeled = np.setdiff1d(tr, labeled)\n\n    def fit(a, b):\n        return ExtraTreesClassifier(\n            n_estimators=96,\n            min_samples_leaf=8,\n            class_weight="balanced",\n            random_state=seed,\n            n_jobs=2,\n        ).fit(a, b)\n\n    initial = fit(x[labeled], y[labeled])\n    p = initial.predict_proba(x[unlabeled])[:, 1]\n    accepted = np.abs(p - 0.5) > 0.42\n    expanded = fit(\n        np.concatenate([x[labeled], x[unlabeled][accepted]]),\n        np.r_[y[labeled], p[accepted] >= 0.5],\n    )\n    output = {}\n    for name, model in [("labeled_only", initial), ("pseudo_labels", expanded)]:\n        prob = model.predict_proba(x[te])[:, 1]\n        output[name] = {\n            **scores(y[te], prob),\n            "balanced_accuracy": float(balanced_accuracy_score(y[te], prob >= 0.5)),\n            "test_subjects": len(np.unique(g[te])),\n        }\n    save((initial, expanded, np.unique(g[te]), int(accepted.sum())), f"models-{seed}")\n    return output\n\n\nif __name__ == "__main__":\n    runs = {s: run(s) for s in SEEDS}\n    first = next(iter(runs.values()))\n    means = {\n        k: {m: float(np.mean([r[k][m] for r in runs.values()])) for m in v}\n        for k, v in first.items()\n    }\n    (ROOT / "results.json").write_text(\n        json.dumps({"seeds": SEEDS, "runs": runs, "means": means}, indent=2)\n    )\n    print(means)\n',encoding='utf8')
subprocess.run([sys.executable,'train.py'],cwd=root,check=True)
```
