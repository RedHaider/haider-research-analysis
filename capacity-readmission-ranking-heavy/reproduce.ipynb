# Readmission ranking under a fixed outreach budget (simulation)

Compare ranking by prior inpatient visits with a gradient-boosted ranker using eight encounter measurements. Hold out 25% of people in each split. Measure the fraction of positive encounters captured in the top 10% and the positive fraction within that budget. Generate two encounters per simulated patient with overlapping predictors and patient-level frailty.

```python
from pathlib import Path
import subprocess, sys
root = Path.cwd() / 'capacity-readmission-ranking-heavy-run'
root.mkdir(exist_ok=True)
subprocess.run([sys.executable,'-m','pip','install',*['numpy==2.2.6', 'scipy==1.15.3', 'scikit-learn==1.7.2', 'joblib==1.6.0']],check=True)
```

```python
(root / 'train.py').write_text('import json\nimport os\nfrom pathlib import Path\n\nimport joblib\nimport numpy as np\nfrom sklearn.metrics import (\n    average_precision_score,\n    roc_auc_score,\n)\n\nROOT = Path(__file__).resolve().parent\nDATA = Path(os.environ.get("PORTFOLIO_DATA", ROOT / "data"))\nSEEDS = [13, 37, 61]\n\n\ndef save(obj, name):\n    folder = ROOT / "checkpoints"\n    folder.mkdir(exist_ok=True)\n    joblib.dump(obj, folder / (name + ".joblib"), compress=3)\n\n\ndef scores(y, p):\n    return {\n        "auroc": float(roc_auc_score(y, p)),\n        "average_precision": float(average_precision_score(y, p)),\n    }\n\n\ndef dataset(seed):\n    from scipy.special import expit\n\n    rng = np.random.default_rng(seed)\n    groups = np.repeat(np.arange(6500), 2)\n    frailty = np.repeat(rng.normal(size=6500), 2)\n    x = np.c_[\n        rng.poisson(np.exp(0.2 + 0.4 * frailty)), rng.normal(size=(len(groups), 7))\n    ]\n    p = expit(\n        -2.7\n        + 0.37 * x[:, 0]\n        + 0.9 * x[:, 1]\n        - 0.6 * x[:, 2]\n        + 0.4 * x[:, 1] * x[:, 2]\n        + 0.55 * frailty\n    )\n    return x, rng.binomial(1, p), groups\n\n\nfrom sklearn.ensemble import HistGradientBoostingClassifier\nfrom sklearn.model_selection import GroupShuffleSplit\n\n\ndef run(seed):\n    x, y, groups = dataset(seed)\n    train, test = next(\n        GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=seed).split(\n            x, y, groups\n        )\n    )\n    assert not set(groups[train]) & set(groups[test])\n    model = HistGradientBoostingClassifier(\n        max_iter=130, max_leaf_nodes=15, l2_regularization=2, random_state=seed\n    ).fit(x[train], y[train])\n    save(model, f"ranker-{seed}")\n    out = {}\n    for method, p in [\n        ("prior_inpatient_visits", x[test, 0]),\n        ("learned_priority", model.predict_proba(x[test])[:, 1]),\n    ]:\n        order = np.argsort(-p, kind="stable")\n        eligible = order[: int(0.10 * len(test))]\n        out[method] = {\n            **scores(y[test], p),\n            "recall_at_budget": float(y[test][eligible].sum() / y[test].sum()),\n            "precision_at_budget": float(y[test][eligible].mean()),\n        }\n    return out\n\n\nif __name__ == "__main__":\n    runs = {s: run(s) for s in SEEDS}\n    first = next(iter(runs.values()))\n    means = {\n        k: {m: float(np.mean([r[k][m] for r in runs.values()])) for m in v}\n        for k, v in first.items()\n    }\n    (ROOT / "results.json").write_text(\n        json.dumps({"seeds": SEEDS, "runs": runs, "means": means}, indent=2)\n    )\n    print(means)\n',encoding='utf8')
subprocess.run([sys.executable,'train.py'],cwd=root,check=True)
```
